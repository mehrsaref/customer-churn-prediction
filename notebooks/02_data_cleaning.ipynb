```python
from pathlib import Path
import pandas as pd

RAW = Path("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
PROCESSED = Path("../data/processed/telco_churn_clean.csv")

df = pd.read_csv(RAW)
df.shape
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

#errors="coerce" means if it's not a number turn it to NaN.
```

```python
print(df["TotalCharges"].dtype)
print(df["TotalCharges"].isna().sum())
```

```python
df[df["TotalCharges"].isna()][["customerID","tenure","MonthlyCharges","Churn","TotalCharges"]]
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
df.isna().sum().sum()
```

```python
df["Churn"] = df["Churn"].map({"Yes":1, "No":0})
df["Churn"].value_counts(normalize=True)* 100
```

```python
df = df.drop(columns=["customerID"])
df.shape
```

```python
PROCESSED.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(PROCESSED, index=False)
print(f"Saved {df.shape[0]} rows x {df.shape[1]} columns to {PROCESSED}")

#Make sure the processed-data folder exists,
# ,save my DataFrame there as a CSV without the Pandas index
# , and tell me how many rows and columns I saved.
```
